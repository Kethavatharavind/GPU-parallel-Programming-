```python
%%writefile assigment1_1.cu
#include <iostream>
#include<cuda_runtime.h>
using namespace std;

int main(){
    cudaDeviceProp p;
    cudaGetDeviceProperties(&p,0);

    cout<< "GPU Name : "<<p.name<<endl;
    cout<< "Gobal memory : "<<p.totalGlobalMem/(1024*1024)<<"MB"<<endl;
    cout<< "shared memory per block : "<<p.sharedMemPerBlock/(1024*1024)<<"MB"<<endl;
    cout<< "max threads per Block : "<<p.maxThreadsPerBlock<<endl;
    cout<< "warp Size : "<<p.warpSize<<endl;
    cout<<"Register per block : "<<p.regsPerBlock<<endl;
    cout<<"no of SM : "<<p.multiProcessorCount<<endl;
    cout<<"max threads per SM : "<<p.maxThreadsPerMultiProcessor<<endl;
    cout<<"L2 cache size : "<<p.l2CacheSize/(1024*1024)<<"MB"<<endl;

    return 0;
}
```

```python
!nvcc assigment1_1.cu -o q1
!./q1
```

```python
%%writefile assigment1_2.cu
#include<iostream>
#include<cuda_runtime.h>
using namespace std;

__global__ void kernel(){

  __shared__ int count;
  int tid=threadIdx.x;
  if(tid==0)count=0;

  __syncthreads();
  printf("threads id : %d\n",tid);
  atomicAdd(&count,1);

  __syncthreads();
}


int main(){
  kernel<<<1,8>>>();
  cudaDeviceSynchronize();
  return 0;
}
```

```python
!nvcc assigment1_2.cu -o q2
!./q2
```
